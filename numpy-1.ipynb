# Array Dimension

```python
import numpy as np
#one dimensional array
d1 = np.array([1,2,3,4,5])
d1# one-dimensional
```

```python
#2d Array
d2 = np.array([[1,2,3],[4,5,6]])
d2
```

```python
np.zeros((2, 2))    # 3x3 array of zeros
np.ones((2, 4))     # 2x4 array of ones
np.full((2, 2), 7)  # 2x2 array filled with 7
np.eye(4)           # 4x4 identity matrix
np.arange(1, 10, 2) # [1, 3, 5, 7, 9] (from first value 1 to 10 increasing by 2)
np.linspace(0, 1, 5) # [0. 0.25 0.5 0.75 1.] (evenly spaced)
```

# Indexing & Slicing

```python
import numpy as np
#indexing
arr = np.array([10, 20, 30, 40, 50, 60, 70])
print(arr[0])  
print(arr[-1])
```

```python
print(arr[2::2])  #taking every 2 vlaues
sliced = arr[1:4].copy()  #creating copy of sliced data
sliced[0] = 11             #slice provides view only not duplicate
print(arr)
print(sliced)
```

```python
arrr = np.array([1,4,5,6,7,5,4,5,8,7,])
inx = [1,9,7]     #selecting from these indexes
print(arrr[inx])
```

# Boolean Masking

```python
ar = np.array([1,4,5,6,7,5,4,5,8,7,])
ar[ar>5]  #printing greater value then 5
```
