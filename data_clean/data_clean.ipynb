```python
with open("insta_data.txt", encoding='utf-8') as f:
    data = f.read()
    #breaking user data into individual profile
chunks = data.split("\n\n")
chunks = [item for item in chunks if len(item)>3]

#storing individual profile data in standard format
def store_data(chunk):

    chunk = chunk.strip()    #removing space from both side
    lines = chunk.split("\n")

    # Initializing all fields to empty string to handle missing data 
    user_name = ""
    name = ""
    post = ""
    follower = ""
    following = ""
    bio = ""

    if len(lines) > 0:
        user_name = lines[0]
    if len(lines) > 1:
        name = lines[1]
    if len(lines) > 2:
        post = lines[2]
    if len(lines) > 3:
        follower = lines[3]
    if len(lines) > 4:
        following = lines[4]
    if len(lines) > 5:
        bio = lines[5:]

    return {"username": user_name, "name": name, "post": post, "follower": follower, "following": following, "bio": bio}
#printing all data
all_chunks = []
for chunk in chunks:
    parsed_chunk = store_data(chunk)
    all_chunks.append(parsed_chunk )
print(all_chunks)

#Finding profile with minimum posts and minimum followers

min_post = min(all_chunks, key = lambda x:x["post"])
print("\n""Minimim post:--", min_post)

min_follower = min(all_chunks, key= lambda y:y["follower"])
print("\n""Minimim post:--", min_follower)
```

#Creating json from above file

```python
import json
s = json.dumps(all_chunks, indent = 4)
print(s)
with open("final_data.json", "w") as f:
    f.write(s)
```
