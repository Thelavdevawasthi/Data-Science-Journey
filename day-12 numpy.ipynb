```python
import numpy as np
```

```python
# vectorization
arr = np.array([1,2,3,4,5,6])
result = arr**2
print(result)
```

```python
#Broadcasting
ar = np.array([1,2,3,4,5,6 
result = ar+10    #10 added to all element of array
print(result)
```

```python
#broadcasting 2d array into 1d array
a1 = np.array([[1,2,3],[4,5,6]])
a2 = np.array([1,2,3])
result = a1 + a2     #this [1,2,3] is added to all element  of 2d array
# --- broadcasting a2 to a1
print(result)
```

```python
# Simulating a dataset (5 samples, 3 features)
data = np.array([[10, 20, 30],
                 [15, 25, 35],
                 [20, 30, 40],
                 [25, 35, 45],
                 [30, 40, 50]])
mean = data.mean(axis=0)
std = data.std(axis=0)
normalized_form = (data - mean)/std
print(normalized_form)
```

# Mathematical functions in Numpy

```python
arr = np.array([2,4,6,8,10])
np.mean(arr)     #calculating mean
np.std(arr)      #calculating standard deviation
np.var(arr)       #calculating variance
np.argmin(arr)    #index of minimun value
np.argmax(arr)    #index of max value
```
